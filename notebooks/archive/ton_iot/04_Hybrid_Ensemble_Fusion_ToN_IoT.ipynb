# SENTRi-X: Hybrid Ensemble Fusion & Multimode Evaluation
Evaluates RF, CNN, and Hybrid fusion on the frozen holdout split with full provenance and prediction evidence.
Validation data is strictly isolated from test holdout.

```python
import sys
from pathlib import Path
ROOT_DIR = Path("..").resolve()
if str(ROOT_DIR) not in sys.path:
    sys.path.insert(0, str(ROOT_DIR))

import numpy as np
from sentrix_ml.adapters.ton_iot import load_ton_iot
from sentrix_ml.splits import stratified_split
from sentrix_ml.provenance import sampling_audit
from sentrix_ml.balancing import balance_training_rows
from sentrix_ml.preprocessing import PreprocessingPipeline
from sentrix_ml.training import train_rf, train_cnn
from sentrix_ml.inference import hybrid_predict, predict_rf, predict_cnn
from sentrix_ml.evaluation import compute_multimode_metrics, save_prediction_evidence

print("Imports ready.")
```

```python
# Step 1: Prepare data and train models
X, y, info = load_ton_iot(ROOT_DIR / "data" / "raw" / "ton_iot", sample_n=2000, seed=42)
X_train, X_val, X_test, y_train, y_val, y_test, manifest = stratified_split(
    X,
    y,
    test_fraction=0.20,
    val_fraction=0.10,
    seed=42,
    domain="ton_iot", metadata=info["metadata"], require_class_support=True,
    source_file_hashes=info["source_file_hashes"], sampling_metadata=sampling_audit(info),
)

pipeline = PreprocessingPipeline().fit(X_train)
X_train_scaled = pipeline.transform(X_train)
X_val_scaled = pipeline.transform(X_val)
X_test_scaled = pipeline.transform(X_test)
X_fit, y_fit, balancing_audit = balance_training_rows(X_train_scaled, y_train, seed=42)

# Train Random Forest on train partition
rf = train_rf(X_fit, y_fit, random_state=42)
p_rf = predict_rf(rf, X_test_scaled)
print(f"Trained RF on {len(X_train)} samples. Mean RF test attack prob: {p_rf.mean():.4f}")
```

```python
# Step 2: Train CNN with validation set (X_val, y_val) - NEVER test set
try:
    import tensorflow as tf
    cnn, history = train_cnn(X_fit, y_fit, X_val_scaled, y_val, epochs=2, batch_size=64, verbose=0)
    p_cnn = predict_cnn(cnn, X_test_scaled)
    print(f"Trained CNN. Mean CNN test attack prob: {p_cnn.mean():.4f}")
except ImportError:
    print("TensorFlow not installed. Using baseline probability for demonstration.")
    p_cnn = np.full_like(p_rf, 0.5)

p_hybrid = (p_rf + p_cnn) / 2.0
eval_res = compute_multimode_metrics(
    y_test.to_numpy(dtype=int),
    p_rf=p_rf,
    p_cnn=p_cnn,
    p_hybrid=p_hybrid,
    domain="ton_iot",
    run_type="smoke",
)
print("\nMultimode Evaluation Results:")
for mode, m in eval_res.modes.items():
    print(f"Mode [{mode:6s}]: Acc={m['accuracy']:.4f}, Prec={m['precision']:.4f}, Rec={m['recall']:.4f}, F1={m['f1']:.4f}")
```
